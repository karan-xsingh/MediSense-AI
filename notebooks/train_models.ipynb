```python
# Load dataset
df3 = pd.read_csv('../datasets/kidney_disease.csv')

# Drop id column
df3 = df3.drop('id', axis=1)

# Clean whitespace from column names
df3.columns = df3.columns.str.strip()

# Replace tab+question mark values with NaN
df3 = df3.replace(r'^\s*\??\s*$', np.nan, regex=True)

# Strip whitespace from string columns
for col in df3.columns:
    if df3[col].dtype == 'object':
        df3[col] = df3[col].str.strip()

# Fill missing values with mode
df3 = df3.fillna(df3.mode().iloc[0])

# Encode ALL categorical/object columns including target
le = LabelEncoder()
for col in df3.columns:
    if df3[col].dtype == 'object':
        df3[col] = le.fit_transform(df3[col].astype(str))

# Force all columns to numeric
df3 = df3.apply(pd.to_numeric, errors='coerce')
df3 = df3.fillna(0)

# Features and target
X3 = df3.drop('classification', axis=1)
y3 = df3['classification']

# Split
X_train3, X_test3, y_train3, y_test3 = train_test_split(X3, y3, test_size=0.2, random_state=42)

# Train
kidney_model = RandomForestClassifier(n_estimators=100, random_state=42)
kidney_model.fit(X_train3, y_train3)

# Accuracy
y_pred3 = kidney_model.predict(X_test3)
print(f"Kidney Disease Model Accuracy: {accuracy_score(y_test3, y_pred3)*100:.2f}%")

# Save model
joblib.dump(kidney_model, '../models/kidney_model.pkl')
print("✅ Kidney model saved!")
```
